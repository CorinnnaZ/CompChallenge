```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
import seaborn as sb # gives access to the heatmap function itself
```

```python
## this code will be using a function as its input which is different to my intention
## but a good introduction to this idea
```

```python
def resolution_function(resolution):
    x = np.linspace(0,10,resolution)
    y = np.linspace(0,10,resolution)

    x,y = np.meshgrid(x,y)

    def f(A,B):
        return np.sin(np.sqrt(A**2 + B**2))

    z = f(x,y)

    ax = sb.heatmap(z)
    ax.invert_yaxis()
```

```python
interact(resolution_function, resolution=(1,1000,1))
```
